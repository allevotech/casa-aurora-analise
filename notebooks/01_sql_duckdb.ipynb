**Problema:** em 2025 o faturamento cresceu, mas a margem caiu. Onde a Casa Aurora está perdendo dinheiro?

Rode as duas células de preparação (Shift+Enter) e responda às 8 perguntas. Faturamento = pedidos com `status = 'delivered'`. Receita do item = `quantity * unit_price - discount`; custo = `quantity * unit_cost`.

```python
# Onde estão os dados: se a pasta data/clean existir (repositório baixado), usa ela;
# senão (Google Colab), lê direto do repositório-modelo no GitHub. Não precisa mudar nada.
from pathlib import Path

DATA_BASE_URL = "https://raw.githubusercontent.com/allevotech/casa-aurora-analise/main/data/clean"
CANDIDATES = [Path("data/clean"), Path("../data/clean"), Path("../../data/clean")]
local_dir = next((p for p in CANDIDATES if (p / "orders.csv").exists()), None)
BASE = str(local_dir) if local_dir else DATA_BASE_URL
print("Lendo dados de:", BASE)
```

```python
try:
    import duckdb
except ImportError:  # o Colab normalmente já tem; se não tiver, instala
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "duckdb"], check=True)
    import duckdb

con = duckdb.connect()
# Datas ficam como texto (AAAA-MM-DD): o mesmo SQL roda igual no SQLite e no MySQL.
DATE_COLS = {"orders": ["order_date"], "customers": ["signup_date"]}
for table in ["customers", "products", "orders", "order_items"]:
    types = ", ".join(f"'{c}': 'VARCHAR'" for c in DATE_COLS.get(table, []))
    opts = f", types={{{types}}}" if types else ""
    con.execute(f"CREATE TABLE {table} AS SELECT * FROM read_csv('{BASE}/{table}.csv', header=true{opts})")
    print(table, con.execute(f"SELECT COUNT(*) FROM {table}").fetchone()[0], "linhas")


def sql(query):
    """Roda uma consulta e devolve uma tabela (DataFrame) para visualizar."""
    linhas = [l.strip() for l in query.splitlines()]
    if all(not l or l.startswith("--") for l in linhas):
        print("✍️ Escreva sua consulta SQL entre as aspas triplas e rode a célula de novo.")
        return None
    return con.sql(query).df()
```
